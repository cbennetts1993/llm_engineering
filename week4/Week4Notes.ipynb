#### Chinchilla Scaling Law

Coined by google deepmind team

related to how you think about nnumber of parameters in a model

how many you have is ~ proportional to the number of training data

If you're seeing diminishing returns from training more data, than this 'law' gives you a rule of thumb for scaling your model.

similarly, if you double your model parameters, your model needs more training data

### Limitations of Benchmarks

- not consistently applied
- too narrow in scope
- hard to measure nuanced reasoning
- training data leakage
- overfitting = tweaking for specific parameters can cause it to be really good at that, but horribly at other tasks or when presented the task in a different way.

Possibly:
- Frontier LLMs may be 'aware' that they are being evaluated.

# Week 4 Challenge

Build a product that converts python code to C++ to improve code performance 

prompt: "Please reimplement this python code in C++ with the fastest possible implementation for an M! Mac. Only respond with the C++ code. Do not explain your implementation. The only requirement is that the C++ code prints the same results and runs fast."

```python
# Test code

import time

def calculate(iterations, param1, param2):
    result = 1.0
    for i in range(1, iterations+1):
        j = i * param1 - param2
        result -= (1/j)
    return result

start_time = time.time()
result = calculate(100_000_000, 4, 1) * 4
end_time = time.time()

print(f"Result: {result:.12f}")
print(f"Execution Time: {(end_time - start_time):.6f}")
```

## How to evaluate performance?

### Two kinds of performance metrics
#### Model Centric or Technical metrics
- Loss (eg. cross-entropy loss) (-log(p))
- Perplexity (e ^ -log(p)) how many tokens needed to predict the next token.
- Accuracy 
- Precision, Recall, F1
- AUC-ROC

Easiest to Optimize

#### Business-centric or Outcome metrics
- KPIs tied to business objectives
- ROI
- improvements in time, cost, resources
- Customer satisfaction
- Benchmark comparisons

Most Tangible impacts
