#### Chinchilla Scaling Law

Coined by google deepmind team

related to how you think about nnumber of parameters in a model

how many you have is ~ proportional to the number of training data

If you're seeing diminishing returns from training more data, than this 'law' gives you a rule of thumb for scaling your model.

similarly, if you double your model parameters, your model needs more training data



### 7 common benchmarks
| Benchmark    | what's being evaluated?    |Description     |
| ------------- | ------------- | ----------- |
| ARC| Reasoning | A benchmark for evaluating scientific reasoning, multiple-choice questions |
| DROP | Language Comprehension | Distill details from text then add, count or sort |
| HellaSwag | Common Sense | "Hard Endings, Long Contexts and Low Shot Activities" |
| MMLU | Understanding | Factual recall, reasoning and problem solving across 57 subjects |

### 3 more specific benchmarks

| Benchmark    | what's being evaluated?    |Description     |
| ------------- | ------------- | ----------- |
| ELO | Chat | results from head-to-head face-offs with other LLMs, as with ELO in chess |
| HumanEval | Python Coding | 164 problems writing code based on docstrings |
| MultiPL-E | Broader Coding | Translation of HumanEval to 18 programming languages |

### Limitations of Benchmarks

- not consistently applied
- too narrow in scope
- hard to measure nuanced reasoning
- training data leakage
- overfitting = tweaking for specific parameters can cause it to be really good at that, but horribly at other tasks or when presented the task in a different way.

Possibly:
- Frontier LLMs may be 'aware' that they are being evaluated.
