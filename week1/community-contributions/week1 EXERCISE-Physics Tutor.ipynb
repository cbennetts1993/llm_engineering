and responds with an explanation. This is a tool that you will be able to use yourself during the course!

````python
# imports
import os
import requests
from typing import List
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
````

````python
# constants

open_ai = OpenAI()
ollama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
MODEL_QWEN = 'qwen3:14b'
````

````python
# set up environment

api_key = os.getenv("OPENAI_API_KEY")

load_dotenv(override=True)

if api_key and api_key.startswith('sk-proj-') and len(api_key) > 10:
    print("API KEY looks good so far")
else:
    print("API key looks invalid. Please see the troubleshooting documentation")


system_prompt = "You are an assistant that answers technical questions \
                from university level students. You are able to provide clear and detailed \
                explanations on the queried subject. You should also be transparent and \
                tell the student which llm model you are before you answer the question."


def handle_query(query: str):

    handled_query = "Here is a query from one of your students"
    handled_query += "Provide them with a detailed explanation to this question: \_n"
    handled_query += query
    
    return handled_query
````

````python
# here is the question; type over this to ask something new

question = """
How would a cosmologist go about estimating the number of stars in the universe?
"""
````

````python
# Get gpt-4o-mini to answer, with streaming

def query_gpt4(query: str, streaming: bool = True):
    stream = open_ai.chat.completions.create(
        model=MODEL_GPT,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": handle_query(query)}
        ],
        stream=streaming
    )
    
    if streaming:
        response = ""
        display_handle = display(Markdown(""), display_id=True)
        for chunk in stream:
            response += chunk.choices[0].delta.content or ""
            response = response.replace("```","").replace("markdown", "")
            update_display(Markdown(response), display_id=display_handle.display_id)
    else:
        response = stream.choices[0].message.content
        display(Markdown(response))

query_gpt4(query=question, streaming=True)
````
